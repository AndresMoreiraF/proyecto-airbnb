```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

TINTA, ACENTO, MORADO, GRIS = "#1B1A2F", "#E8963C", "#4C3F91", "#8A879E"

# Cargar y limpiar precio
df = pd.read_csv("../datos/listings.csv.gz")
df["precio"] = df["price"].str.replace("[$,]", "", regex=True).astype(float)
df = df[df["precio"] > 0]

# Selección de columnas
cols = ["room_type", "property_type", "accommodates", "bedrooms", "beds", "bathrooms",
        "neighbourhood_cleansed", "latitude", "longitude",
        "host_is_superhost", "host_response_rate", "host_acceptance_rate",
        "host_listings_count", "hosts_time_as_host_years",
        "number_of_reviews", "review_scores_rating", "reviews_per_month",
        "minimum_nights", "availability_365", "precio"]
df = df[cols].copy()

# Estadísticas
print(df.describe().round(1))
print("\n% faltantes:")
print(df.isna().mean().mul(100).round(1).sort_values(ascending=False).to_string())

# Gráficas
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0,0].hist(df["precio"], bins=80, color=MORADO, edgecolor="white")
axes[0,0].set_title("Precio por noche")

axes[0,1].hist(np.log1p(df["precio"]), bins=80, color=ACENTO, edgecolor="white")
axes[0,1].set_title("log(1 + precio)")

df.groupby("room_type")["precio"].median().sort_values().plot.barh(ax=axes[1,0], color=MORADO)
axes[1,0].set_title("Precio mediano por tipo")

corr = df.select_dtypes("number").corr()["precio"].drop("precio").sort_values()
corr.plot.barh(ax=axes[1,1], color=[MORADO if v>0 else ACENTO for v in corr])
axes[1,1].set_title("Correlación con precio")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

TINTA, ACENTO, MORADO, GRIS = "#1B1A2F", "#E8963C", "#4C3F91", "#8A879E"

df = pd.read_csv("../datos/listings.csv.gz")
df["precio"] = df["price"].str.replace("[$,]", "", regex=True).astype(float)
df = df[(df["precio"] > 0) & (df["accommodates"] > 0)]
df["precio_huesped"] = df["precio"] / df["accommodates"]

# Cortar outliers extremos para ver la distribución real
p99 = df["precio"].quantile(0.99)
dfh = df[df["precio"] <= p99]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(dfh["precio"], bins=80, color=MORADO, edgecolor="white", alpha=0.9)
axes[0].axvline(dfh["precio"].median(), color=ACENTO, ls="--", lw=2,
                label=f'Mediana: ${dfh["precio"].median():,.0f} COP')
axes[0].set(title="Precio por noche", xlabel="COP", ylabel="Frecuencia")
axes[0].legend()

ph99 = dfh["precio_huesped"].quantile(0.99)
dfh2 = dfh[dfh["precio_huesped"] <= ph99]

axes[1].hist(dfh2["precio_huesped"], bins=80, color=ACENTO, edgecolor="white", alpha=0.9)
axes[1].axvline(dfh2["precio_huesped"].median(), color=MORADO, ls="--", lw=2,
                label=f'Mediana: ${dfh2["precio_huesped"].median():,.0f} COP')
axes[1].set(title="Precio por noche / huésped", xlabel="COP", ylabel="Frecuencia")
axes[1].legend()

plt.tight_layout()
plt.show()
```
